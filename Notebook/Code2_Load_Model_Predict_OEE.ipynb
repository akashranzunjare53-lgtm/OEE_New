### Load Saved Model and Predict Tomorrow OEE

#### Import Library 

```python
import pickle
import pandas as pd
```

#### Load saved model

```python
with open("../Model/OEE_Tomorrow_Model.pkl", "rb") as file:
    saved_data = pickle.load(file)

loaded_model = saved_data["model"]
saved_features = saved_data["features"]

print("Model loaded successfully!")

print("Features used by model:")
print(saved_features)
```

#### --> Enter yesterday's machine data 

```python
latest_machine_data = pd.DataFrame({
    "Machine_Age": [1],
    "OEE_Lag_1": [72.17],
    "Avail_Hrs_Lag_1": [8.18],
    "MC_Running_Hrs_Lag_1": [7.48],
    "Total_DT_Lag_1": [0.7],
    "Prod_Qty_Lag_1": [422],
    "Rej_Qty_Lag_1": [21],
    "Maintenance_Flag_Lag_1": [1]  # Yes = 1 No = 0
})
```

#### --> Predict tomorrow's OEE

```python
tomorrow_oee = loaded_model.predict(latest_machine_data[saved_features])

print("Predicted OEE for tomorrow:")

print(round(tomorrow_oee[0], 2),"%"
)
```
